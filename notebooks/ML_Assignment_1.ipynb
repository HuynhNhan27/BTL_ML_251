# **Huấn luyện mô hình**
Sau khi đã tiền xử lý và chuẩn bị dữ liệu, nhóm tiến hành huấn luyện một số mô hình học máy để tìm ra thuật toán phù hợp nhất cho bài toán dự đoán giá sản phẩm. Nhóm đã lựa chọn và thử nghiệm các mô hình sau: Linear Regression, RandomForestRegressor, SVR và MLPRegressor.

- Quá trình Huấn luyện:
    - Với mỗi mô hình, nhóm đã huấn luyện trên tập dữ liệu huấn luyện (X_train, y_train) đã được chuẩn hóa.
    - Sau khi huấn luyện, mô hình được sử dụng để dự đoán giá sản phẩm trên tập dữ liệu kiểm tra X_test.
    - Do biến mục tiêu y_test đã được chuẩn hóa và biến đổi logarit trước đó, nhóm đã phải thực hiện biến đổi ngược (y_scaler.inverse_transform và np.expm1) để đưa các giá trị dự đoán về thang đo ban đầu, giúp việc đánh giá trở nên ý nghĩa hơn.
- Đánh giá Hiệu suất:
    - Để đánh giá hiệu suất của mỗi mô hình, nhóm sử dụng ba chỉ số phổ biến cho bài toán hồi quy:
      - Mean Squared Error (MSE): Đo lường mức độ sai lệch trung bình của các dự đoán. Giá trị càng thấp càng tốt.
      - Root Mean Squared Error (RMSE): Thể hiện mức độ sai lệch trung bình giữa giá trị dự đoán và giá trị thực tế theo cùng đơn vị với biến mục tiêu. RMSE nhạy cảm hơn với các sai số lớn do đặc tính bình phương trước khi lấy căn. Giá trị RMSE càng thấp thì mô hình càng chính xác.
      - Mean Absolute Error (MAE): Đo lường độ sai lệch tuyệt đối trung bình. Giá trị này dễ hiểu hơn MSE vì nó cùng đơn vị với biến mục tiêu. Giá trị càng thấp càng tốt.
      - R-squared (R2): Thể hiện mức độ phù hợp của mô hình với dữ liệu, với giá trị càng gần 1 càng tốt.

```python
# Scaling y_train for better training
y_scaler = tabular.Log1pRobustScaler()
y_train = pd.DataFrame(y_scaler.fit_transform(y_train.values))
```

```python
X_train = preprocessing_pipeline.fit_transform(X_train)
X_test = preprocessing_pipeline.transform(X_test)
```

```python
from sklearn.model_selection import ParameterGrid
# Model chosing
param_grid = [
    {
        "model": [LinearRegression()],
        "model__fit_intercept": [True],
        "model__copy_X": [True],
        "model__n_jobs": [None]
    },
    {
        "model": [RandomForestRegressor()],
        "model__n_estimators": [100],
        "model__criterion": ["squared_error"],
        "model__max_depth": [None],
        "model__min_samples_split": [2],
        "model__min_samples_leaf": [1],
        "model__n_jobs": [None],
        "model__random_state": [42]
    },
    {
        "model": [SVR()],
        "model__kernel": ["rbf"],
        "model__degree": [3],
        "model__gamma": ["scale"],
        "model__C": [1.0],
        "model__epsilon": [0.1]
    },
    {
        "model": [MLPRegressor()],
        "model__hidden_layer_sizes": [(100,)],
        "model__activation": ["relu"],
        "model__solver": ["adam"],
        "model__alpha": [0.0001],
        "model__batch_size": ["auto"],
        "model__learning_rate": ["constant"],
        "model__max_iter": [200],
        "model__random_state": [42]
    }
]
grid = ParameterGrid(param_grid=param_grid)
```

```python
results = []

for params in grid:
    # Tạo pipeline với model tương ứng
    pipe = Pipeline([
        # ("preprocess", preprocessing_pipeline),
        ("model", params["model"])
    ])
    pipe.set_params(**params)
    pipe.set_output(transform="pandas")

    pipe.fit(X_train, y_train)
    y_pred_scaled = pipe.predict(X_test)
    y_pred = pd.DataFrame(y_scaler.inverse_transform(y_pred_scaled.reshape(-1, 1)))

    mse = mean_squared_error(y_test, y_pred)
    rmse = np.sqrt(mse)
    mae = mean_absolute_error(y_test, y_pred)
    r2  = r2_score(y_test, y_pred)

    # Lưu kết quả
    results.append({
        "params": params["model"].__class__.__name__,
        "mse": mse,
        "rmse": rmse,
        "mae": mae,
        "r2": r2
    })

# Xuất kết quả ra DataFrame
df_results = pd.DataFrame(results)
print(df_results)
```

### **Nhận xét về kết quả thử nghiệm các cấu hình tiền xử lý (Preprocessing Configs)**

Dựa trên kết quả chạy thử nghiệm với nhiều cấu hình tiền xử lý khác nhau (Imputation, Scaling, PCA, Encoding), ta có những nhận xét sau:

1.  **Linear Regression:**
    *   Mô hình cho kết quả rất tệ (R2 âm) trên hầu hết các cấu hình không sử dụng Scaler là log1p_robust, khi sử dụng log1p_robust, kết quả thu được khá tốt (R2 khoảng 0.81 - 0.82).
    *   Điều này cho thấy mối quan hệ giữa các đặc trưng và giá sản phẩm (`discounted_price`) có tính phi tuyến cao, hoặc dữ liệu chứa nhiều nhiễu/outlier mà mô hình tuyến tính đơn giản không thể xử lý tốt, cần phải có các bước tiền xử lý để đưa đặc trưng về dạng tuyến tính.

2.  **Random Forest Regressor:**
    *   Đây là mô hình hoạt động tốt nhất và ổn định nhất.
    *   Hiệu suất rất cao (R2 > 0.99) ngay cả khi không thực hiện scaling (`scale='none'`).
    *   Việc sử dụng PCA (`pca_0.95` hoặc `pca_0.99`) làm giảm đáng kể hiệu suất (R2 giảm xuống ~0.77), cho thấy việc giữ nguyên các đặc trưng gốc quan trọng hơn là giảm chiều dữ liệu đối với mô hình cây quyết định này.
    *   Encoding: Cả `OneHot` và `Ordinal` đều cho kết quả tốt tương đương nhau.

3.  **SVR (Support Vector Regression):**
    *   Mô hình này cực kỳ nhạy cảm với việc chuẩn hóa dữ liệu (Scaling).
    *   Khi không scale (`scale='none'`), mô hình hoàn toàn không hoạt động (R2 âm rất lớn).
    *   Khi sử dụng `StandardScaler` hoặc `RobustScaler`, hiệu suất tăng vọt lên mức rất tốt (R2 ~ 0.95).
    *   Encoding: `OneHot` encoding hoạt động tốt hơn hẳn so với `Ordinal` encoding đối với SVR.

4.  **MLP Regressor (Neural Network):**
    *   Tương tự SVR, MLP cũng yêu cầu dữ liệu phải được chuẩn hóa để thuật toán tối ưu (Adam) hoạt động hiệu quả.
    *   Kết quả tốt nhất đạt được khi kết hợp `StandardScaler` hoặc `MinMaxScaler` với `PCA` và `OneHot` encoding (R2 ~ 0.97).
    *   PCA có vẻ hỗ trợ tốt cho MLP trong việc trích xuất đặc trưng và giảm nhiễu.

**👉 Chiến lược chọn Config cho bước Tuning Hyperparameter:**
